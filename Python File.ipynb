```python
import matplotlib.pyplot as plt

condition_count = df['Medical Condition'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(condition_count.index, condition_count.values, color='grey')
plt.title('Patients by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Number of patients')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

# What is the Age Distribution of Patients ?

```python
plt.figure(figsize=(8, 5))
plt.hist(df['Age'],bins = 20, color='grey', edgecolor='white')
plt.title('Age Distribution of Patients')
plt.xlabel('Age')
plt.ylabel('Number of patients')
plt.tight_layout()
plt.show()
```

# How has the Average treatment Cost changed over Time ?

```python
Average_Cost =df['Billing Amount'].mean()
Average_Cost
```

```python
df['Admission_Month'] = df['Date of Admission'].dt.to_period('M')

```

```python
df.head()
```

```python
Average_Montly_Cost = df.groupby('Admission_Month')['Billing Amount'].mean()
Average_Montly_Cost
```

```python
plt.figure(figsize=(8, 5))
plt.plot(Average_Montly_Cost.index.astype(str), Average_Montly_Cost.values, color='darkgrey',marker='o', markersize=3)
plt.title('Average Treatment Cost by Month')
plt.xlabel('Month')
plt.ylabel('Average Billing Amount')
plt.xticks(rotation=90)
plt.show()
```

# Is there a relationshpi between Patient Age and Treatment Cost ?

```python
senior_avg_cost = df.loc[df['Age']>=60,'Billing Amount'].mean()
senior_avg_cost
```

```python
sample_df = df.sample(1000,random_state =1)
plt.figure(figsize=(8, 5))
plt.scatter(sample_df['Age'],sample_df['Billing Amount'],color='darkgrey',alpha=0.5,s=15)
plt.title('Age vs Treatment Cost')
plt.xlabel('Age')
plt.ylabel('Billing Amount')
plt.tight_layout()
plt.show()
```
